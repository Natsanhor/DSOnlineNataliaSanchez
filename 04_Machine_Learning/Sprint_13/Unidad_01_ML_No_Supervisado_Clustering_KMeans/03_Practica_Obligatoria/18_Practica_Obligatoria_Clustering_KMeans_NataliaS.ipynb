### 1.6

Aquí viene la parte del león. A continuación, agrupa las imágenes utilizando K-Means sobre el dataset de train reducido en el ejercicio anterior. Emplea el método del máximo de score de silueta para obtener el mejor k, probando con K de 5 en 5 hasta 150. ¿Cuál es el k que proporciona el mejor score de silueta? NOTA: Emplea todas las features (no hace falta seleccionar, y ya están escaladas entre 0 y 1)

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

k_values = range(5,151,5)

scores = []

for k in k_values:

    model = KMeans(n_clusters=k, random_state=42)

    labels = model.fit_predict(X_train_pca)

    score = silhouette_score(X_train_pca, labels)

    scores.append(score)

    print("k =", k, "silhouette =", round(score,4))
````

### #1.7

Repite el ejercio anterior empleando ahora el método del codo de Inercia. ¿Sale algo más concluyente o que refuerce el anterior resultado?

````python
inertias = []

for k in k_values:

    model = KMeans(n_clusters=k, random_state=42)

    model.fit(X_train_pca)

    inertias.append(model.inertia_)

plt.figure(figsize=(8,5))

plt.plot(k_values, inertias, marker="o")

plt.xlabel("Número de clusters")
plt.ylabel("Inercia")

plt.title("Método del codo")

plt.show()
````

### #1.8

Quédate con el k obtenido con el método del score de silueta y asigna el modelo con ese k a una variable `best_model`

````python
best_k = k_values[np.argmax(scores)]

print("Mejor número de clusters:", best_k)

best_model = KMeans(
    n_clusters=best_k,
    random_state=42
)

best_model.fit(X_train_pca)
````

### #2.3 EXTRA VOLUNTARIO

Busca el número de clusters k que generen un algoritmo kmeans que a su vez sus distancias a los centroides sean las features de un clasificador RandomForest y que permita al clasificador obtener el mejor rendimiento: ¿Qué rendimiento puedes alcanzar? (en terminos de Accuracy)

````python
k_values = range(5,151,5)

accuracies = []

for k in k_values:

    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

    kmeans.fit(X_train_pca)

    X_train_red = kmeans.transform(X_train_pca)
    X_val_red = kmeans.transform(X_val_pca)

    rf = RandomForestClassifier(
        n_estimators=150,
        random_state=42
    )

    rf.fit(X_train_red, y_train)

    y_pred = rf.predict(X_val_red)

    acc = accuracy_score(y_val, y_pred)

    accuracies.append(acc)

    print("k =",k,"accuracy =",round(acc,4))
````

````python
best_k = k_values[np.argmax(accuracies)]

best_acc = max(accuracies)

print("Mejor k:",best_k)
print("Mejor accuracy:",best_acc)
````
